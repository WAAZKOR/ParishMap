#Imports

```python
from bs4 import BeautifulSoup
import csv
import re
import os
```

#Define Functions

```python

def parse_parishes_html(html_content):
    """
    Parses HTML content containing parish information and extracts data.

    Args:
        html_content: A string containing the HTML data.

    Returns:
        A list of dictionaries, where each dictionary represents a parish
        and contains its extracted information.
    """
    soup = BeautifulSoup(html_content, 'html.parser')
    parishes_data = []

    # Find all div elements with the class 'output_parish'
    parish_divs = soup.find_all('div', class_='output_parish')

    for parish_div in parish_divs:
        parish_info = {}

        # Extracting basic information
        parish_info['title'] = extract_text(parish_div, 'span', 'parish_title')
        parish_info['jurisdiction'] = extract_text(parish_div, 'p', 'parish_jurisdiction')
        parish_info['jurcode'] = extract_text(parish_div, 'p', 'parish_jurcode')
        parish_info['address'] = extract_text(parish_div, 'p', 'parish_address')
        parish_info['city'] = extract_text(parish_div, 'span', 'parish_city')
        parish_info['state'] = extract_text(parish_div, 'span', 'parish_state')
        parish_info['zip'] = extract_text(parish_div, 'span', 'parish_zip')
        parish_info['geocode'] = extract_text(parish_div, 'span', 'parish_geocode')
        parish_info['latitude'] = extract_text(parish_div, 'span', 'parish_latitude')
        parish_info['longitude'] = extract_text(parish_div, 'span', 'parish_longitude')

        # Extracting information from the 'parish_more_info' div
        more_info_div = parish_div.find('div', class_='parish_more_info')
        if more_info_div:
            parish_info['phone'] = extract_text(more_info_div, 'span', 'parish_phone')
            parish_info['fax'] = extract_text(more_info_div, 'span', 'parish_fax')
            parish_info['website'] = extract_text(more_info_div, 'span', 'parish_website')
        else:
            parish_info['phone'] = ''
            parish_info['fax'] = ''
            parish_info['website'] = ''

        parishes_data.append(parish_info)

    return parishes_data

def extract_text(parent_element, tag, class_name):
    """
    Finds an element within a parent element by tag and class and returns its text.

    Args:
        parent_element: The BeautifulSoup element to search within.
        tag: The HTML tag of the element to find.
        class_name: The class name of the element to find.

    Returns:
        The text content of the found element, or an empty string if not found.
    """
    found_element = parent_element.find(tag, class_=class_name)
    if found_element:
        return found_element.get_text(strip=True)
    return ''

def write_to_csv(data, filename):
    """
    Writes a list of dictionaries to a CSV file.

    Args:
        data: A list of dictionaries containing the data to write.
        filename: The name of the CSV file to create.
    """
    if not data:
        print("No data to write to CSV.")
        return

    keys = data[0].keys()
    with open(filename, 'w', newline='', encoding='utf-8') as output_file:
        dict_writer = csv.DictWriter(output_file, fieldnames=keys)
        dict_writer.writeheader()
        dict_writer.writerows(data)
```

#Use Pandas to combine all csvs and then remove duplicates

```python
import pandas as pd

# Specify the path to the folder containing your CSV files
folder_path = '/content/csvs'  # Replace with the actual path to your folder

# Create an empty list to store DataFrames
all_data = []

# Iterate through all files in the specified folder
for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):  # Only process files ending with .csv
        file_path = os.path.join(folder_path, filename)
        try:
            df = pd.read_csv(file_path)
            all_data.append(df)
            print(f"Successfully read: {filename}")
        except FileNotFoundError:
            print(f"Error: File not found - {file_path}")
        except Exception as e:
            print(f"Error reading {file_path}: {e}")

# Concatenate all DataFrames in the list into a single DataFrame
if all_data:
    combined_df = pd.concat(all_data, ignore_index=True)

    # Remove duplicate rows
    final_df = combined_df.drop_duplicates()

    # Save the result to a new CSV file
    final_df.to_csv('combined_and_deduplicated.csv', index=False)

    print("Successfully combined CSVs, removed duplicates, and saved to 'combined_and_deduplicated.csv'")
else:
    print(f"No CSV files found in the folder: {folder_path}")
```
